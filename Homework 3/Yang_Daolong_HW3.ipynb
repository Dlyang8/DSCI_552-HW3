#### ii. Extraction

```python
data_features = []

for i in bending_1_len:
    data = pd.read_csv(f"AReM/bending1/dataset{i}.csv", skiprows = 4, usecols= range(1, 7))
    features = data.describe().drop(index = "count").unstack().to_frame().T
    data_features.append(features)

for i in bending_2_len:
    if i == 4:
        data = pd.read_csv(f"AReM/bending2/dataset{i}.csv", skiprows=5, sep=r"\s+", header=None, names=column_names, usecols=range(1, 7))
    else:
        data = pd.read_csv(f"AReM/bending2/dataset{i}.csv", skiprows=4, usecols=range(1, 7))
    features = data.describe().drop(index="count").unstack().to_frame().T
    data_features.append(features)

rest = ['cycling', 'lying', 'sitting', 'standing', 'walking']
for i in rest:
    for j in rest_len:
        data = pd.read_csv(f"AReM/{i}/dataset{j}.csv", skiprows = 4, usecols= range(1, 7))
        features = data.describe().drop(index = "count").unstack().to_frame().T 
        data_features.append(features)

data_features = pd.concat(data_features, ignore_index = True)
data_features  
```

#### iii. Standard Deviation

```python
std = data_features.std()
std
```

```python
estimated_std = data_features.std()

bootstrap_std = []

for i in range(10000):
    bootstrap_sample = data_features.sample(n=len(data_features), replace=True, random_state=i)
    bootstrap_std.append(bootstrap_sample.std())

bootstrap_std = pd.DataFrame(bootstrap_std)

bootstrap_results = pd.DataFrame({
    "Standard Deviation": estimated_std,
    "90% CI Lower": bootstrap_std.quantile(0.05),
    "90% CI Upper": bootstrap_std.quantile(0.95)
})

bootstrap_results = bootstrap_results.round(3)
bootstrap_results
```

### (a) Linear Train

When the relationship is linear we can expect that the training RSS for a more linear regression model to be higher than the cubic regression model. For models with linear relationship, training RSS will not increase as model flexibility decreases. As the cubic regression model is more flexible than the linear regression model, the cubic regression model would then would have a lower RSS training score than the linear regression model. 

### (b) Linear Test

When the relationship between X and Y is linear, the expectation is that the test RSS will be higher for a more flexible model. With the linear regression model since it is a less flexible model, the test RSS will be lower than a more flexible model like the cubic regression model. As such, the cubic regression model for test RSS will be higher than the linear regression model test RSS. 

### (c) Not Linear Train

When the relationship between X and Y is non linear, the expectation is that the training RSS for a less flexible model would be higher than a more flexible model. Since the linear regression model is less flexible than the cubic regression model, the training RSS for the linear regression model would be higher than the training RSS for a cubic regression model. 

### (d) Not Linear Testing

In this scenario due to the unknown nature of how non-linear the model is, more information is required to state which model would test RSS be higher for. If the true relationship is substantially nonlinear, the cubic regression model may have the lower test RSS because of its lower bias. However, if the relationship is only slightly nonlinear, the linear model may have the lower test RSS because it has lower variance.
